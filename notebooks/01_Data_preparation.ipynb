#  Identify potential outliers

I used the IQR method to identify unusually high or low prices and freight charges. It flagged 8,427 rows for price and 12,134 rows for freight value. These records require review before any cleaning decision.

```python
Q1 = order_items["price"].quantile(0.25)
Q3 = order_items["price"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

order_items[
    (order_items["price"] < lower) |
    (order_items["price"] > upper)
]
```

```python
Q1 = order_items["freight_value"].quantile(0.25)
Q3 = order_items["freight_value"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

order_items[
    (order_items["freight_value"] < lower) |
    (order_items["freight_value"] > upper)
]
```
